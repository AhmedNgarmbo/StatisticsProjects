 Analyzing corporate financial ratios is crucial for several reasons, as it helps investors, analysts, and other stakeholders assess the 
 financial health, performance, and viability of a company. 


 Evaluating Profitability
 Profit Margins: Ratios like Net Profit Margin and Gross Profit Margin help assess how efficiently a company is converting its revenue into 
 profits. Higher margins suggest better efficiency in managing costs.
 Return on Equity (ROE) and Return on Assets (ROA): These ratios indicate how well a company is using its equity and assets to generate profits. 
 A higher ROE and ROA are typically signs of effective management and strong financial performance.

    
 Assessing Liquidity
 Current Ratio and Quick Ratio measure a company’s ability to cover its short-term obligations with its short-term assets. Analyzing liquidity 
ratios helps investors understand if the company can meet its immediate financial commitments, which is critical for short-term financial stability.
Cash Flow Ratios: These ratios analyze the company's ability to generate cash from its operations to pay bills, fund growth, and provide returns 
# to shareholders.


 Understanding Leverage and Risk
 Debt-to-Equity Ratio: This ratio helps investors assess how much debt a company is using to finance its operations. Companies with high debt 
relative to equity may face higher financial risks, especially in times of economic downturns.
# Interest Coverage Ratio: Shows how easily a company can pay interest on its outstanding debt. A low ratio might indicate potential difficulty 
# in covering interest payments.


# Measuring Efficiency
# Asset Turnover Ratio: Measures how efficiently a company uses its assets to generate revenue. A high asset turnover ratio suggests effective 
# utilization of assets.
# Inventory Turnover Ratio: Indicates how quickly a company sells and replaces its inventory. High turnover rates suggest strong sales or 
# efficient inventory management.

    
# Valuation and Market Sentiment
# Price-to-Earnings (P/E) Ratio: Used to assess if a company’s stock is overvalued or undervalued by comparing its current share price to its earnings 
# per share (EPS). A high P/E ratio may suggest that the stock is overpriced, or it could indicate strong growth expectations from investors.
# Price-to-Sales (P/S) and Price-to-Book (P/B) Ratios: These ratios are useful for evaluating companies that do not yet have significant profits,
# providing alternative measures of value.

                                                                                                                                              
# Comparative Analysis
# Financial ratios allow for comparison across companies within the same industry, helping investors gauge which companies are performing better 
# relative to their peers.
# By looking at a company’s trends over time, analysts can assess whether its financial performance is improving or deteriorating, and predict 
# future trends.

                                                                                                                               
# Forecasting Financial Health and Sustainability
# Ratios help investors predict the company’s future growth, profitability, and ability to weather economic challenges. For instance, a company 
# with strong liquidity, profitability, and low debt levels is likely to be more resilient during an economic downturn.

                                                                                                                               
# Credit and Loan Risk Assessment
# Lenders use financial ratios to assess a company's creditworthiness. Ratios like the Debt Ratio and Interest Coverage Ratio are vital in 
# determining whether a company can repay its loans and if it's a good candidate for credit.


# Strategic Decision Making
# Companies themselves use financial ratios to make informed decisions. They can identify areas for improvement (e.g., reducing operating costs,
# managing debt more efficiently) and adjust their strategies to optimize financial performance.


# Investor Confidence
# Ratios provide transparency into a company's financial performance and health, which helps build investor confidence. Investors are more likely 
# to invest in companies with strong, transparent financial health and sound performance metrics.


```python
import requests
from bs4 import BeautifulSoup
from pandas import DataFrame
import itertools
import numpy as np
from itertools import chain
import time
import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

 
url_base = "https://finviz.com/quote.ashx?t="

tckr = ['UNH','CRM','VZ','V','WBA','WMT','DIS','PG','TRV','AXP','AMGN','AAPL','BA','CAT','CSCO','CVX','GS','HD','HON','IBM',
        'INTC','JNJ','KO','JPM','MCD','MMM','MRK','MSFT','NKE']
     
i = 1

url_list = [(s, url_base + s) for s in tckr]
data_list = []

 
headers = {'User-Agent': 'Mozilla/5.0 (X11; Ubuntu; Linux x86_64; rv:76.0) Gecko/20100101 Firefox/76.0'}


import re

def clean_numeric_string(s):
    # Replace '-' with '0'
    s = s.replace('-', '0')
    # Remove any non-numeric characters
    s = re.sub(r'[^0-9]', '', s)
    return s
    
 
for t, url in url_list:
    print(i)
    i += 1
    print(t, url)
    print(f'Scraping ticker {t}...')
    
    soup = BeautifulSoup(requests.get(url, headers=headers).content, 'html.parser')
    
    l = []
    for row in soup.select('.snapshot-table2 tr'):
        l.append([td.text for td in row.select('td')])

    x = list(chain.from_iterable(l))
    
    # Ensure we start with a fresh dictionary
    d = {}
    if len(x) >= 2:  # Check that there's valid data to parse
        d = dict(zip(x[::2], x[1::2]))
    
    d['Index'] = t  # Ensure the ticker symbol is recorded
    
    data_list.append(d)  # Append the new dictionary for each ticker

    # Slow things down so too many requests (too frequently) don't get blocked
    time.sleep(2)



# Function to convert percentages
def convert_percentage(value):
    if isinstance(value, str) and '%' in value:
        return float(value.replace('%', '').strip()) / 100
    return value


# Function to convert magnitude values (B and M)
def convert_magnitude(value):
    if isinstance(value, str):
        value = value.strip().upper()
        if 'B' in value:
            return float(value.replace('B', '').strip()) * 1e9
        elif 'M' in value:
            return float(value.replace('M', '').strip()) * 1e6
    return value


    
df = pd.DataFrame(data_list)

display(df.head())
```
