 Analyzing corporate financial ratios is crucial for several reasons, as it helps investors, analysts, and other stakeholders assess the 
 financial health, performance, and viability of a company. 


 Evaluating Profitability
 Profit Margins: Ratios like Net Profit Margin and Gross Profit Margin help assess how efficiently a company is converting its revenue into 
 profits. Higher margins suggest better efficiency in managing costs.
 Return on Equity (ROE) and Return on Assets (ROA): These ratios indicate how well a company is using its equity and assets to generate profits. 
 A higher ROE and ROA are typically signs of effective management and strong financial performance.

    
 Assessing Liquidity
Current Ratio and Quick Ratio measure a company’s ability to cover its short-term obligations with its short-term assets. Analyzing liquidity 
ratios helps investors understand if the company can meet its immediate financial commitments, which is critical for short-term financial stability.
Cash Flow Ratios: These ratios analyze the company's ability to generate cash from its operations to pay bills, fund growth, and provide returns 
to shareholders.


Understanding Leverage and Risk
Debt-to-Equity Ratio: This ratio helps investors assess how much debt a company is using to finance its operations. Companies with high debt 
relative to equity may face higher financial risks, especially in times of economic downturns.
Interest Coverage Ratio: Shows how easily a company can pay interest on its outstanding debt. A low ratio might indicate potential difficulty 
in covering interest payments.


Measuring Efficiency
Asset Turnover Ratio: Measures how efficiently a company uses its assets to generate revenue. A high asset turnover ratio suggests effective 
utilization of assets.
Inventory Turnover Ratio: Indicates how quickly a company sells and replaces its inventory. High turnover rates suggest strong sales or 
efficient inventory management.

    
Valuation and Market Sentiment
Price-to-Earnings (P/E) Ratio: Used to assess if a company’s stock is overvalued or undervalued by comparing its current share price to its earnings 
per share (EPS). A high P/E ratio may suggest that the stock is overpriced, or it could indicate strong growth expectations from investors.
Price-to-Sales (P/S) and Price-to-Book (P/B) Ratios: These ratios are useful for evaluating companies that do not yet have significant profits,
providing alternative measures of value.

                                                                                                                                              
Comparative Analysis
Financial ratios allow for comparison across companies within the same industry, helping investors gauge which companies are performing better 
relative to their peers.
By looking at a company’s trends over time, analysts can assess whether its financial performance is improving or deteriorating, and predict 
future trends.

                                                                                                                               
 Forecasting Financial Health and Sustainability
 Ratios help investors predict the company’s future growth, profitability, and ability to weather economic challenges. For instance, a company 
 with strong liquidity, profitability, and low debt levels is likely to be more resilient during an economic downturn.

                                                                                                                               
Credit and Loan Risk Assessment
Lenders use financial ratios to assess a company's creditworthiness. Ratios like the Debt Ratio and Interest Coverage Ratio are vital in 
determining whether a company can repay its loans and if it's a good candidate for credit.


Strategic Decision Making
Companies themselves use financial ratios to make informed decisions. They can identify areas for improvement (e.g., reducing operating costs,
managing debt more efficiently) and adjust their strategies to optimize financial performance.


Investor Confidence
Ratios provide transparency into a company's financial performance and health, which helps build investor confidence. Investors are more likely 
to invest in companies with strong, transparent financial health and sound performance metrics.


```python
import requests
from bs4 import BeautifulSoup
from pandas import DataFrame
import itertools
import numpy as np
from itertools import chain
import time
import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

 
url_base = "https://finviz.com/quote.ashx?t="

tckr = ['UNH','CRM','VZ','V','WBA','WMT','DIS','PG','TRV','AXP','AMGN','AAPL','BA','CAT','CSCO','CVX','GS','HD','HON','IBM',
        'INTC','JNJ','KO','JPM','MCD','MMM','MRK','MSFT','NKE']
     
i = 1

url_list = [(s, url_base + s) for s in tckr]
data_list = []

 
headers = {'User-Agent': 'Mozilla/5.0 (X11; Ubuntu; Linux x86_64; rv:76.0) Gecko/20100101 Firefox/76.0'}


import re

def clean_numeric_string(s):
    # Replace '-' with '0'
    s = s.replace('-', '0')
    # Remove any non-numeric characters
    s = re.sub(r'[^0-9]', '', s)
    return s
    
 
for t, url in url_list:
    print(i)
    i += 1
    print(t, url)
    print(f'Scraping ticker {t}...')
    
    soup = BeautifulSoup(requests.get(url, headers=headers).content, 'html.parser')
    
    l = []
    for row in soup.select('.snapshot-table2 tr'):
        l.append([td.text for td in row.select('td')])

    x = list(chain.from_iterable(l))
    
    # Ensure we start with a fresh dictionary
    d = {}
    if len(x) >= 2:  # Check that there's valid data to parse
        d = dict(zip(x[::2], x[1::2]))
    
    d['Index'] = t  # Ensure the ticker symbol is recorded
    
    data_list.append(d)  # Append the new dictionary for each ticker

    # Slow things down so too many requests (too frequently) don't get blocked
    time.sleep(2)



# Function to convert percentages
def convert_percentage(value):
    if isinstance(value, str) and '%' in value:
        return float(value.replace('%', '').strip()) / 100
    return value


# Function to convert magnitude values (B and M)
def convert_magnitude(value):
    if isinstance(value, str):
        value = value.strip().upper()
        if 'B' in value:
            return float(value.replace('B', '').strip()) * 1e9
        elif 'M' in value:
            return float(value.replace('M', '').strip()) * 1e6
    return value


    
df = pd.DataFrame(data_list)

display(df.head())
```

```python
df.replace('-', 0, inplace=True)


# Apply percentage and magnitude conversions to the dataframe
df['ROA'] = df['ROA'].apply(convert_percentage)
df['ROE'] = df['ROE'].apply(convert_percentage)
df['Sales Y/Y TTM'] = df['Sales Y/Y TTM'].apply(convert_percentage)
df['Income'] = df['Income'].apply(convert_magnitude)
df['Market Cap'] = df['Market Cap'].apply(convert_magnitude)


display(df.tail())
```

```python
# Define the thresholds for each ratio (bad, ok, good)

metric_thresholds = {
    "P/E": [10, 25, 35],
    "EPS (ttm)": [5, 10, 15],
    "Forward P/E": [5, 20, 30],
    "PEG": [0, 2, 3],
    "P/S": [0, 1, 5],
    "P/B": [0, 1, 3],
    "P/FCF": [0, 10, 25],
    "Beta": [0, 1, 2],
    "Quick Ratio": [0, 1, 2],
    "RSI (14)": [10, 50, 90],
    "Debt/Eq": [0, 0.5, 1],
    "LT Debt/Eq": [0, 0.5, 1],
    "ROA": [.05, .15, .25],
    "ROE": [.50, .15, .25],
    "Sales Y/Y TTM": [.05, .10, .20]
}

    
# Function to assign score based on thresholds
def assign_score(value, thresholds):
    if value is None or value == '':
        return 0  # Treat missing values as 0 score
    try:
        value = float(value)
    except (ValueError, TypeError):
        return 0  # Treat unconvertible values as 0

    if value < thresholds[1]:
        return 0  # Bad
    elif value < thresholds[2]:
        return 5  # Ok
    else:
        return 10  # Good


# Function to evaluate each stock's metrics
# Function to evaluate each stock's metrics
def evaluate_stock_metrics(df, metric_thresholds):
    stock_scores = {}

    for ticker in df['Index'].unique():  # Ensure unique stock tickers
        stock_data = df[df['Index'] == ticker]

        if stock_data.empty:
            continue  # Skip if no data for this ticker

        total_score = 0
        metrics_count = 0

        print(f"Evaluating stock: {ticker}")

        for metric, thresholds in metric_thresholds.items():
            if metric not in df.columns:
                print(f"Warning: Missing column {metric}")
                continue  # Skip if column does not exist in df

            metric_value = stock_data[metric].values[0]

            # Convert metric_value to a float safely
            try:
                metric_value = float(metric_value)
            except (ValueError, TypeError):
                continue  # Skip if non-numeric or missing

            # Assign score
            score = assign_score(metric_value, thresholds)

            # Print individual score for the metric
            print(f"  {metric}: {metric_value} -> Score: {score}")

            total_score += score
            metrics_count += 1

        # Calculate the overall ranking for the stock
        overall_score = total_score / metrics_count if metrics_count > 0 else 0
        stock_scores[ticker] = overall_score

        # Print overall score for the stock
        print(f"Overall score for {ticker}: {overall_score}/10\n")

    return stock_scores

# Example usage:
# Assuming your dataframe is named `df` and contains the financial metrics
metric_scores = evaluate_stock_metrics(df, metric_thresholds)


# Print the overall ranking for each stock
for ticker, score in metric_scores.items():
    print(f"Stock {ticker} Overall Score: {score}/10")
```

```python
# Store results in a list
results = []

for ticker, score in metric_scores.items():
    results.append({"Ticker": ticker, "Score": score})

# Convert to a DataFrame
score_df = pd.DataFrame(results)

# Sort by Score in descending order (best to worst)
score_df = score_df.sort_values(by="Score", ascending=False)

# Display the sorted DataFrame
print(score_df)
```

```python
# Descriptions of top ~15 Financial Ratios!

# 1. P/E (Price-to-Earnings Ratio):
# Description: This ratio compares the price of a company's stock to its earnings per share (EPS). It gives an indication of how much investors are 
# willing to pay for each dollar of earnings.        
# Good Ratio: A P/E ratio around 15-20 is often considered fair or reasonably priced for many companies.
# Bad Ratio: A very high P/E (e.g., 100+) may indicate overvaluation, while a very low P/E (e.g., below 5) might signal that the company is undervalued 
# or facing significant problems.

    
# 2. EPS (Earnings Per Share) (ttm):
# Description: EPS is the portion of a company’s profit allocated to each outstanding share of common stock, calculated over the trailing twelve 
# months (ttm). It is used to measure the profitability of a company.
# Good Ratio: Higher EPS typically indicates a more profitable company. For instance, an EPS of $5 is a good sign of solid earnings.
# Bad Ratio: Negative or declining EPS (e.g., -$1) may signal poor financial health and profitability issues.


# 3. Forward P/E (Forward Price-to-Earnings Ratio):
# Description: This is a variant of the P/E ratio, using projected earnings instead of past earnings. It gives an estimate of how much investors 
# are willing to pay for expected future earnings.
# Good Ratio: A forward P/E ratio of 10-20 is often considered good, depending on the industry and the company's growth prospects.
# Bad Ratio: A forward P/E above 50 can be a red flag, indicating overvaluation based on overly optimistic earnings projections.


# 4. PEG (Price/Earnings to Growth Ratio):
# Description: The PEG ratio takes into account the company's P/E ratio and its expected earnings growth rate. It helps assess whether a stock 
# is overvalued or undervalued based on its growth rate.
# Good Ratio: A PEG ratio around 1 suggests a company is fairly valued relative to its earnings growth.
# Bad Ratio: A PEG ratio above 2 might indicate overvaluation, while a very low PEG (below 0.5) could suggest undervaluation or declining growth.


# 5. P/S (Price-to-Sales Ratio):
# Description: This ratio compares a company’s stock price to its revenue per share. It’s useful for evaluating companies with little or no earnings.
# Good Ratio: A P/S ratio below 2 is typically considered a good value, especially for growing companies.
# Bad Ratio: A very high P/S (e.g., above 10) could indicate that the company’s stock is overvalued relative to its sales.

        
# 6. P/B (Price-to-Book Ratio):
# Description: This ratio compares a company’s market value to its book value (assets minus liabilities). It helps assess whether a stock is 
# under or overvalued based on its net assets.
# Good Ratio: A P/B ratio around 1 or below suggests the stock is fairly valued or undervalued.
# Bad Ratio: A very high P/B (e.g., 5 or above) might indicate that the stock is overpriced.


# 7. P/FCF (Price-to-Free Cash Flow Ratio):
# Description: This ratio compares the market price of a company to its free cash flow per share, indicating how much investors are willing to 
# pay for each dollar of free cash flow.
# Good Ratio: A P/FCF ratio below 15 is often seen as a good value.
# Bad Ratio: A very high P/FCF (e.g., 30+) could indicate overvaluation or poor cash flow performance.

                                                                                                                                                    
# 8. Beta:
# Description: Beta measures a stock’s volatility in relation to the market. A beta of 1 indicates that the stock moves with the market, while a 
# beta higher than 1 indicates greater volatility.
# Good Ratio: A beta of 0.8-1.2 is ideal for investors seeking moderate risk.
# Bad Ratio: A beta significantly higher than 1 (e.g., 2.0) may indicate high risk, while a very low beta (e.g., 0.2) could signal low volatility.


# 9. Quick Ratio:
# Description: This ratio assesses a company’s ability to meet its short-term liabilities with its most liquid assets (excluding inventories). 
# It’s a measure of liquidity.
# Good Ratio: A quick ratio of 1 or higher is generally considered healthy.
# Bad Ratio: A quick ratio below 1 indicates that the company may have trouble covering its short-term liabilities with its liquid assets.


# 10. RSI (14):
# Description: The Relative Strength Index (RSI) measures the speed and change of price movements to identify overbought or oversold conditions. 
# The 14 refers to the 14-day period used.
# Good Ratio: An RSI around 30-70 indicates normal trading conditions. An RSI above 70 suggests the stock may be overbought, while below 30 
# suggests it may be oversold.
# Bad Ratio: RSI above 70 may signal a potential price pullback (overbought), while below 30 could signal a potential rebound (oversold).


# 11. Debt/Eq (Debt-to-Equity Ratio):
# Description: This ratio compares a company’s total debt to its shareholders' equity, indicating the level of financial leverage.
# Good Ratio: A debt-to-equity ratio below 1 is generally considered safe, showing that the company is not overly reliant on debt.
# Bad Ratio: A high debt-to-equity ratio (e.g., above 2) could indicate that the company is highly leveraged and may be at risk in case of 
# financial difficulties.


# 12. LT Debt/Eq (Long-Term Debt-to-Equity Ratio):
# Description: This ratio focuses on a company’s long-term debt compared to equity, showing the long-term financial risk.
# Good Ratio: A ratio below 0.5 is considered healthy, indicating low long-term financial risk.
# Bad Ratio: A ratio above 1.0 indicates a higher reliance on long-term debt, which could pose risk if interest rates rise or earnings decrease.


# 13. ROA (Return on Assets):
# Description: This ratio measures a company’s ability to generate profit from its assets.
# Good Ratio: A high ROA (e.g., above 5%) indicates efficient use of assets to generate profit.
# Bad Ratio: A very low ROA (e.g., below 1%) may indicate that the company is inefficient in utilizing its assets.

            
# 14. ROE (Return on Equity):
# Description: ROE measures how effectively a company is using its equity to generate profit.
# Good Ratio: An ROE above 15-20% is considered excellent, indicating efficient use of shareholders' equity.
# Bad Ratio: An ROE significantly below 10% might indicate poor profitability and inefficient capital use.

    
# 15. Sales Y/Y TTM (Sales Year-over-Year, Trailing Twelve Months):
# Description: This ratio measures the percentage change in sales compared to the same period in the previous year.
# Good Ratio: A positive sales growth (e.g., 10% YoY growth) is a good indicator of business expansion.
# Bad Ratio: Negative sales growth (e.g., -5%) may signal a decline in the business's market position or operational issues.
```
