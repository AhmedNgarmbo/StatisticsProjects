```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(123)

salary = np.random.normal(
    loc=70000,
    scale=15000,
    size=200
)

df = pd.DataFrame({
    "salary": salary
})

df.head()
```

```python
df.describe()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["salary"], bins=20)

plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.title("Salary Distribution")

plt.show()
```
